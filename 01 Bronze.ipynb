```python
%sql
CREATE OR REPLACE TABLE small_projects.finops.bronze_billing_usage AS
SELECT * FROM read_files('/Volumes/small_projects/finops/raw/billing_usage.csv', format => 'csv', header => true);

CREATE OR REPLACE TABLE small_projects.finops.bronze_resources AS
SELECT * FROM read_files('/Volumes/small_projects/finops/raw/resources.csv', format => 'csv', header => true);

CREATE OR REPLACE TABLE small_projects.finops.bronze_sku_prices AS
SELECT * FROM read_files('/Volumes/small_projects/finops/raw/sku_prices.csv', format => 'csv', header => true);
```

```python
%sql
SELECT 'usage' AS tbl, COUNT(*) FROM small_projects.finops.bronze_billing_usage
UNION ALL SELECT 'resources', COUNT(*) FROM small_projects.finops.bronze_resources
UNION ALL SELECT 'prices', COUNT(*) FROM small_projects.finops.bronze_sku_prices;
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE small_projects.finops.bronze_usage_stream
AS SELECT * FROM STREAM read_files(
  '/Volumes/small_projects/finops/raw/usage_stream/',
  format => 'csv', header => true);

SELECT COUNT(*) FROM small_projects.finops.bronze_usage_stream;
```
